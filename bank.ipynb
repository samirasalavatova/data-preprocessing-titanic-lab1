```python
import pandas as pd 

df = pd.read_csv("bank.csv.xls") 
df.head()
```

```python
education_mode = df['education'].mode()[0]
df['education'] = df['education'].fillna(education_mode)
df.head()
```

```python
df = pd.get_dummies(df, columns=['marital', 'job', 'housing'], drop_first=True)
df.head()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler() 
df['balance'] = scaler.fit_transform(df[['balance']])
df.head()
```

```python
df.to_csv("bank.csv.xls.final", index=False)
```
